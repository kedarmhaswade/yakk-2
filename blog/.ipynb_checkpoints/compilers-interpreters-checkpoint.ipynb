### On Compilers and Interpreters (30 November 2025)
I wrote this quick blog-post while I was reviewing [Python Functions](https://github.com/kedarmhaswade/yakk-2/blob/main/phase-1/day10_pretest.ipynb).

I stumbled upon the following statements as I read the _Functions_ section in _Python In a Nutshell_ by Martelli et.al. (with a few changes for context):

> When the Python **interpreter** (the `python` executable on your computer) encounters a `def` statement in your Python script or REPL, it **compiles** the function body, creating a function object.

This statement subtly describes the dual nature of `python` (the executable), which is both a compiler and an interpreter. It also prompted me to reexamine _compilation_ and _interpretation_ in the context of programming languages. I had always encountered these terms. I simply asked myself: **Do I understand these terms and the difference between them now?** 

This blog-post answers that question.

First, let's look up the _contemporary_ dictionary meanings of the two verbs: _compile_ and _interpret_. I have stated the most relevant meanings in a boldface.

- **Compile (v.)**: 1) get or gather together (synonym: to accumulate), 2) put together out of existing material (e.g. compile a list), or 3) **use a computer program (a "compiler") to translate source code written in a particular programming language into computer-readable machine code that can be executed**.
- **Interpret (v.)**: 1) make sense of; assign meaning to (e.g. interpret a message) 2) give rendition of (e.g. a Beethoven sonata) 3) create an image or likeness of (synonym: to represent) 3) **restate words from one language into another language (synonym: to translate)**, or 4) make sense of a language.

The two most relevant definitions (modified slightly to Python, but it doesn't matter) are 
- Compile: Use the `python` executable to translate a Python program into the Intel microprocessor's machine code (instructions) which can be _run_ or _executed_ (actual physical changes happen at the semi-conductor level!)
- Interpret: Use the `python` executable to translate or restate the words of Python (e.g. `for` statements) into another language like Intel microprocessor's machine code.

Sadly, both these definitions are insufficient in the modern (in the 2000s and beyond) context of computer programs written in programming languages. The above definition of _compile_ was perhaps made in the 1960s or 70s when to compile was solely to translate a statically-typed source language (e.g. C) program directly into the instructions for the machine.
