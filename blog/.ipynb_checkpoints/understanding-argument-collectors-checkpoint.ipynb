### Understanding `*` and `**` in Python Function Signatures

Let's consider the positional argument collector (`*`) in its simplest form: It _collects_ arguments in a tuple. The named argument collector (``**``) is similar, except that it _collects_ arguments into a dictionary.

There are two views of functions:

1) Developer View: Held by the programmer defining the function (signature and body).
2) User View: Held by the programmer calling the function.

In practice, these views (or roles) are often assumed by the same person. 

Let's say the _Developer_ wants to define a function that adds numbers. 

A natural signature for such a function is: 

#1) `def add(a, b)`.

A simple call to this function is: `s = add(3, 4)`.

Since addition is associative, we can also add more than two numbers by _chaining_ function calls: `a + b + c = a + (b + c) = (a + b) + c`. For example, `add(add(3, 4), 5)`.

To offer flexibility, the Developer can accept a list of numbers from the User. Does _adding_ make sense with only one number? In arithmetic, it doesn't; the _addition operation is defined for_ two operands. The Developer may also provide that flexibility -- `b` gets a _default_ value 0: `a = a + 0`. 

Does it make sense to accept no numbers? Yes, going by the same argument: `0 = 0 + 0`; why not set both the parameters to 0 by default?

The Developer, therefore, ponders this signature: 

#2) `def add(a = 0, b = 0)`.

All the following function calls are now valid:

```
add() # => evaluates to 0, but looks odd
# bind x, y
add(x) # => evaluates to x, looks odd
add(x + y) # => evaluates to x + y
```

What if the User has a list of numbers? To use `add`, the user must write their own looping or recursive calls to sum the numbers in the list. 

If the Developer doesn't heed this 'requirement', their `add` function won't be perceived as useful by many Users. They might argue that the Developer is not implementing the abstract procedure conveyed by the name _add_. A function named `add` must be flexible in that it should abstract out the summing process in a variety of situations.

To entertain a list or iterable of numbers, which makes the most sense when the User has more than two numbers to sum (arguably, the most frequent use case), the Developer considers 

#3) `def add(xs = []) # xs: an iterable.`

and _requires_ Users to pass an iterable that defaults to an empty list. Here's how Users call it now:

```
add() # => evaluates to 0, looks odd
add([]) # => evaluates to 0, doesn't look that odd
# bind x, y, z to numbers
add([x]) # => evaluates to x
add([x, y]) # => evaluates to x + y
add([x, y, z]) # => evaluates to x + y + z.
# bind xs to a list or iterable
add(xs) # sums the list members

# TypeError on passing non-iterable types
add(x) 
add(x, y) 
add(x, y, z)
```

This is better; those who want to pass a list are happy. The function logic deals only with an iterable.

However, calls like `add(x, y)` that were previously convenient (especially in rudimentary cases) are now illegal. 

The Developer considers supporting individual numbers and list members as arguments.

The positional _collector_ is now handy.

The Developer considers the signature:

#4) `def add(*numbers)`. 

**The Developer, by the new signature, conveys to the User that the passed arguments will be collected in a tuple.** Python arranges (using a special instruction, `CALL_FUNCTION_EX` in Python 3.6+) for the function to receive a tuple. Since a tuple is iterable, the function's logic can, as before, be programmed for an iterable.

Here's how the User calls it now:

```
add() # => evaluates to 0
add(54, 2) # => evaluates to 56
add(1, 55, 2) # => evaluates to 58
ys = [33, 55, 10]
add(ys) # => TypeError!
add(*ys) # => evaluates to 98, observe the preceding *

````python
def add(*xs):
    s = 0
    for x in xs:
        s += x
    return s

print(add())
print(add(54, 2))
print(add(1, 55, 2))
ys = [33, 55, 10]
# The next call results in a TypeError, because the list is collected as a member of another tuple!
# print(add(ys))

print(add(*ys)) # however, this works, but in a convoluted way
````

However, a problem (transparent to the Python programmer) with definition #4 above is that all the arguments are passed individually and copied into a tuple available to the function body. This is rather too cumbersome (though hidden) if you already have a list or tuple, whose members you want to sum!

Python standard library (as of version 3.14) uses both approaches. For the built-in function `sum`, Users must pass an iterable. For `max`, Python provides two flavors because the function definitions are overloaded. We have two flavors of `max`, one that accepts an iterable and one that accepts individual arguments.
