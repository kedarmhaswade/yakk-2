### On Compilers and Interpreters (30 November 2025)
I wrote this quick blog-post while I was reviewing [Python Functions](https://github.com/kedarmhaswade/yakk-2/blob/main/phase-1/day10_pretest.ipynb).

I stumbled upon the following statements as I read the _Functions_ section in _Python In a Nutshell_ by Martelli et.al. (with a few changes for context):

> When the Python **interpreter** (the `python` executable on your computer) encounters a `def` statement in your Python script or REPL, it **compiles** the function body, creating a function object.

This statement subtly describes the dual nature of `python` (the executable), which is both a compiler and an interpreter. It also prompted me to reexamine _compilation_ and _interpretation_ in the context of programming languages. I had always encountered these terms. I simply asked myself: **Do I understand these terms and the difference between them now?** 

This blog-post answers that question.

First, let's look up the _contemporary_ dictionary meanings of the two verbs: _compile_ and _interpret_. I have stated the meanings most relevant to a computing context in a boldface.

1) **Compile (v.)**: 1) get or gather together (synonym: to accumulate), 2) put together out of existing material (e.g. compile a list), or 3) **use a computer program (a "compiler") to translate source code written in a particular programming language into computer-readable machine code that can be executed**.
2) **Interpret (v.)**: 1) make sense of; assign meaning to (e.g. interpret a message) 2) give rendition of (e.g. a Beethoven sonata) 3) create an image or likeness of (synonym: to represent) 3) **restate words from one language into another language (synonym: to translate)**, or 4) make sense of a language.

Aren't the relevant definitions confusing? Don't they mean the same thing (to translate)? Restating (modified slightly to suit Python, but it doesn't matter) them:

1) **Compile**: Use the `python` executable to translate a complete Python program into the Intel microprocessor's machine code (instructions) which can be _run_ or _executed_ (actual physical changes occur at the semiconductor level!)
2) **Interpret**: Use the `python` executable to translate or restate the words from Python (e.g. an assignment like `x = 20`) into another language like Intel microprocessor's (x86 64) machine code (something [like](http://ref.x86asm.net/coder64.html#xC7) `48 c7 c0 01 00 00 00` encoded in binary). This is like how a human interpreter interprets a foreign president's response for the native speakers to _understand_. 

When a program _runs_ on a computer, the CPU simply executes corresponding instructions one by one. 

**Thus, the CPU is the ultimate interpreter or executor which runs the instruction in its machine code**. The CPU does _not_ translate the program written in computer's RAM in a 'language' like `48 c7 c0 01 00 00 00` into some other language, but it, like a faithful and unerring servant, "directly makes sense of it" by taking actions (thus _interpreting_ it--See meaning #1 of _interpret_ above).

For a Python program to do anything on a physical machine, therefore, we need two things:

1) Our Python program's code **translated** into a binary machine code (instructions for a CPU).
2) A CPU that **interprets** or makes sense of it by taking the instructed actions.

We now have more nuanced definitions:
1) **Compile**: Translate something from one _formal_ language into another. Formal languages, unlike natural languages, have _unambiguous grammars_.
2) **Interpret**: Make sense of (well-defined) instructions in a language by taking actions intended therein.

Until about the 70s (and of course even now), we had a programming language like C in which programmers wrote their programs, a compiler like the `gcc` (which is itself a big program written in C and compiled into a computer's machine code) that _compiled_ those C programs into a computer's machine code, and a machine that _interpreted_ that machine code. A typical workflow was straightforward:

1) *Create* a C program file, `a.c`. This file is not _executable_.
2) *Compile* it: For example, `gcc a.c`. This creates an _executable_ file `a.out` which has all the machine instructions. The compiler ensures that `a.c` is _valid_ C code, and `a.out` is valid machine code.
3) *Run* the executable file: `./a.out` which loads the instructions in computer's memory. We didn't call this step _interpretation_, although the computer makes sense of (interprets) those instructions.

This required that different computers had different compilers, although the C program file was the same. You could give your little C program, `a.c`, to your friend on a floppy, and s/he would compile it on their computer to, say, `a.exe`, and run it.

What is a compiler like `gcc` then? Well, it must be an executable file that has all the instructions for the machine. Its job is to take your C program and produce a corresponding executable. There is of course a [bootstrapping problem](https://bootstrappable.org/) here if `gcc` were _not_ available as a sequence of machine instructions. If `gcc` was the name of the executable program that was created by compiling a C (or other language) source program, then we must compile that source program somehow!

Later, the idea of a writing computer programs for a virtual machine came along. That made the term 'interpreter' more popular. A compiler translated programs from a source language into an instruction language for the virtual machine. The instructions for the virtual machine were called the _bytecode_. 

The compiler created the bytecode. 

An interpreter interpreted (made sense of) the bytecode. 

To finally see the physical changes on a real machine, however, interpreter also ran or executed the actual machine code corresponding to the bytecode. 

Two notable programming languages in this regard are Python and Java. Let's consider Python.
