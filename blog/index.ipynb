# Welcome to an Artificial Blog
I maintain a blog here. The blog posts pertain to my research interests.

Each blog post is a Jupyter notebook.

```python
# This notebook is only for quarto to render from
from pathlib import Path
from IPython.display import display, Markdown
import nbformat

src = Path(".")
notebooks = sorted([nb for nb in src.glob("*.ipynb") if nb.stem != "index"])

for nb in notebooks:
    html_file = nb.with_suffix(".html").name

    # Read notebook
    nb_obj = nbformat.read(nb, as_version=4)

    # Try to get title from first Markdown cell
    title = nb.stem  # fallback
    for cell in nb_obj.cells:
        if cell.cell_type == "markdown":
            # Take the first non-empty line
            for line in cell.source.splitlines():
                line = line.strip()
                if line:
                    # Remove leading '# ' if present
                    title = line.lstrip("# ").strip()
                    break
            break

    display(Markdown(f"- [{title}]({html_file})"))
```
