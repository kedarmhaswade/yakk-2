# Hello, World!

This is my first blog post, powered by Quarto and GitHub Workflow. I am not exactly happy because it took a half day to set all this up. But now that it is all set up, I can hopefully produce some readable blog-posts that are version-controlled. 

## My Workflow
- Keep my phase-wise notebooks (which are expected to be detailed explorations) up-to-date. I decided not to convert them into a website like this one. Instead, they are just _executed notebooks (even with errors to demonstrate something)_ committed to [my GitHub repo](https://github.com/kedarmhaswade/yakk-2).
- Just see the individual notebooks in `phase-1`, `phase-2` etc. **I link my all my phase-star notebooks from my blog using the GH source links. A typical link is https://github.com/kedarmhaswade/yakk-2/blob/main/phase-1/day10_pretest.ipynb and it should go verbatim in a blog-post**.
- That is my shortcut to host the notebooks _without_ any explicit hosting overhead! I think it works for me.
- When I want to write a blog-post, I come to this folder ([`/blog`](https://github.com/kedarmhaswade/yakk-2/tree/main/blog) of my repository) and create **my blog-post also as a notebook (contains mostly Markdown, but may contain some exploration code-blocks too)**.
- I ask `quarto` to render the static HTML from these "notebooks as blog-posts". For the same purpose, I utilize a GitHub workflow for [static HTML](https://github.com/kedarmhaswade/yakk-2/actions).
- After I am done with my blog-post, I render the Quarto website locally and make sure it succeeds and generates the static pages in the `_site` subfolder of my `blog` folder.
- On commit and push, GitHub runs my workflow and pushes the static HTML (my blog-posts) to [this website](https://kedarmhaswade.github.io/yakk-2/) that you are reading it on.
- Simple, eh?
