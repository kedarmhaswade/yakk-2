## 1. History and Community of FreeBSD

### 1.1 Community
- Port Maintainers
- Utility Maintainers (775 programs and 100 libs that make the core FreeBSD distribution)
  - Core must be stable.
  - To bring something new, you must get rid of something else (equivalent) (e.g. LLVM vs GCC)
- Kernel Maintainers
  - Device Driver Maintainers
  - Focus: Don't support every device out there. This is contrasted with Linux, which wants to support a lot of devices. No slapdash driver! No explosion of drivers! 
- Documentation Maintainers
  - Maintain the online handbook
- QA group
  - Automated testing (via a huge number of scripts)
- Release Engineering
  - Important for downstream schedules (e.g. embedded)
- Sysadmins
  - Email/website etc.
- Advocacy group

All these people are committers.

There is an elected body of seven people. They are in charge of the project for two years. Egalitarian process.

# Lesson 2: Introduction to the Kernel


## The Process Abstraction

The Kernel can be thought of
- as something that distributes or runs processes, or
- as a giant I/O switch.

Each process needs or has:
- CPU time
- Asynchronous events
- Memory (remember, memory means both code and data)
- I/O Descriptors

When you type `ls` (a program), its actual output is perhaps a few hundred bytes max. More "work" is needed to make your program into the process, which is torn down once the program finishes. All the bookkeeping for that is gone. Unix introduced the "process abstraction": A process gets created, the "code" for `ls` (or another program) gets put in there, it does what it's supposed to do, and it goes away.

Before Unix, programmers themselves had to perform all the "bookkeeping work" to convert their software "program" into a running process that actually did something. 

Unix provided the breakthrough that commoditized processes.

I/O descriptor abstraction ("get a descriptor to an input or an output device and just read from it or write to it _uniformly_, not caring about the actual device, it will be all handled for you transparently") was another innovation. 
Before Unix, every I/O code was tightly coupled to the actual physical device.

### Subroutine Call as a Syscall?
Is syscall analogous to a library call, or simple subroutine call?

Syscall "traps" into the operating system (kernel).

If a "function call" affects other processes, it must be a syscall. 

The bottom-half of the Kernel is "asynchronous", quite different from the top-half, which is synchronous. This makes debugging the bottom-half more difficult.
