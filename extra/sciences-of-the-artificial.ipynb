## Introduction
John Laird provides an introduction to the third edition. He includes a brief biographical sketch of Herb Simon, mainly focusing on his academic career. Simon worked with excellent colleagues, Allen Newell (his student and then friend and colleague), and Cliff Shaw.

Laird says that he was searching for answers to questions like:
> What was artificial intelligence? How did it compare to natural intelligence in humans? Was it really a science? Was it (and other _artificial sciences_) intrinsically different from natural sciences?

And found that this book provides many answers (I'd say perspectives).

Simon's main premise is that _artificial had real semantic content_, and he explored its meaning. 

Simon chose a definition of AI:
> He began with a dictionary definition: "**Produced by art rather than by  nature; not genuine or natural; affected; not about the essence of the matter**," choosing the first definition. He refined it \[the definition of AI\] and extended it
> - to _systems designed by humans to attain goals, and to function_, and
> - to entities that adapt to goals and their environment.

This presupposes a definition of _intelligence_. To me, intelligence, as first [defined by John McCarthy and refined by Richard Sutton](http://incompleteideas.net/IncIdeas/DefinitionOfIntelligence.html), suffices:
> **Intelligence is the *computational part of the ability to achieve goals*. A goal-achieving system is more usefully understood in terms of _outcomes_ than in terms of _mechanisms_**.

**KM**: This definition of intelligence draws heavily from the computational theory of mind, which revolves around a central question: **Does the human mind compute?** (and answers it in the affirmative).

Let's take these two definitions of Intelligence and Artificial Intelligence as the starting point.

### Simon's Contributions

Laird writes:
> He \[Simon\] recognized that computers could not only process numbers but also perform symbolic computation and thus perform the same types of reasoning underlying human thought.

And

> What attracted me to Simon and Newell's work was their strategy of developing artificial intelligence systems that closely modeled what we knew of human intelligence, and conversely expanded our knowledge of what was computationally possible for human reasoning. This methodology was their _secret weapon_, which they openly espoused, but it still provided them with unique and novel insights throughout their careers.

> Human behavior data was always an important component of  their theories, possibly best illustrated by the voluminous protocols in _Human Problem Solving_. Their use of knowledge from other fields went beyond psychology, and Simon was unique in the expanse of disciplines he could draw from. But just as important as drawing from many fields was that their theories be computational, and implemented in running computer programs: "In the computer field, the moment of truth is a running program; all else is prophecy."

Laird tells us about Simon and Newell's reception of the Turing Award:

> \[Awarded for\] "Basic contributions to artificial intelligence, the psychology of human cognition, and list processing." **Newell and Simon demystified what it means to think, providing both theory and empirical evidence as to many of the computational processes underlying thought.**"

**KM**: Simon's work clearly presupposes the CTM ([Computational Theory of Mind](https://plato.stanford.edu/entries/computational-mind/)--This resource, Stanford Encyclopedia of Philosophy, is valuable!)

Simon, along with his excellent colleagues, made many contributions across large, independently recognized fields of study and research. I cite below some of those enumerated by Laird. As of 2016, according to Google Scholar, Simon is one of the most highly cited AI/Cognitive Psychology authors:

1) [Why a Diagram is (Sometimes) Worth Ten Thousand Words](https://github.com/kedarmhaswade/yakk-2/tree/main/ref-papers/larkin-simon-why-a-diagram-is-sometimes-worth-10k-words-cognitive-science-jan-mar-1987.pdf)
2) [A Theory of Learning by Doing](https://github.com/kedarmhaswade/yakk-2/tree/main/ref-papers/anzai-simon-theory-of-learning-psychologica-review-v86-n2-1979.pdf)
3) [Protocol Analysis](https://archive.org/details/protocolanalysis0000eric/page/n1/mode/2up): **Use of a psychologist's _subject_'s own verbal reports as data**

**KM**: I wonder what Simon would say of ChatGPT, Gemini, and many other agentic AI systems of 2025! I hope he would feel vindicated ...

Laird returned to this book after a long hiatus (The book has had three editions):
> In returning to this book after many years, it is a pleasure to see once again how Simon distilled the essential ideas from **his lifelong pursuit of understanding organizations and the human mind**.

> One of the joys of reading this book is that you do get to see foundational ideas from your own discipline reflected in others, giving deeper insight into them. Simon's strength as a writer is that a reader does not have to be well-versed in every field that he covers, nor
have an advanced degree in any of those areas.
> He provides building blocks to understand fields as diverse as AI, design, psychology, and economics. My background is in computer science, artificial intelligence, and, to a lesser extent, cognitive psychology. Still, I find all of the topics accessible (and fascinating), including organizational administration, economics, biology, evolution, and design (KM: Might I add _Operations Research_?).


Laird introduces 'Simon's Ant' (wherein Simon thinks of humans to be no more complex than ants):
> I will not attempt to give a comprehensive overview of the book--Simon provides that in the first chapter: "Understanding the Natural and the Artificial Worlds"- but I will explore one of the most memorable aspects of the book, his parable of the ant on the beach, because of its simplicity, and also because of the continued relevance of the underlying ideas it embodies.
> ...
> Simon extends this \[the scenario of an ant walking a 'complex' terrain of a beach\] to human behavior, arguing that the basic structures underlying human behavior 
are actually quite simple--**we are just like the ant, pursuing our goals, attempting to find satisfactory solutions to our current problems, and that the complexity of our behavior is a reflection of the complexity of our tasks, our embodiment, and our environment**:
> >"Human beings, viewed as behaving systems, are quite simple. **The apparent complexity of our behavior over time is largely a reflection of the complexity of the environment in which we find ourselves.**"

**KM**: Such reductionism of complex human beings to simple ants (even though both are navigating a complex terrain) makes many of us uncomfortable (perhaps emotionally). **Are we merely huge goal-seeking ants?** I am still unsure, but it is undeniable that at least a significant part of our--' thinking' people's -- existence is about achieving goals of various kinds ...

**KM**: In our case, we seem to be overly concerned about the finitude of 'our time' here. We even think of death as "a sad reality." Is an ant, which also has biological drives similar to us, aware of passing _Time_? We have no way of knowing its perception of Time. Many of us, however, are keenly aware of Time and want to do more before finally resting. This endeavor compels us to create and use sophisticated tools, optimize, and automate! Is AI humanity's 'ultimate tool'?
