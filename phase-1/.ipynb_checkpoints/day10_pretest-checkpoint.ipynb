## Functions

### Introduction
- A lot has been said about functions.
- In Python, functions always return a result.
- Functions defined within `class` statements are called _methods_.
- In Python, all functions are objects, but not all objects are functions.
- Functions are the only objects that can be _called_ with the function-operator: `(<arguments>)`. To identify this capability, functions are called _callables_ in Python.
- With functions come two things: 1) function definition, and 2) function call.

### Defining Functions
- `def` statement, whose syntax is:
```
def <function-name>(<formal-parameters>):
    <indented-statements>
```


When the `python` interpreter encounters a `def` statement:

> **it compiles the function body**, creating a function object, and binds (or rebinds, if there was an existing binding) `<function_name>` to the compiled function object in the containing namespace (typically the module namespace, or a class namespace when defining methods).

- The function body is not _interpreted_ (only compiled) when the `def` statement is interpreted. 

#### Parameters
My first impression is that Python provides extreme (even confusing) flexibility with parameters. Whoever wants to read others' code needs to become familiar with the breadth provided. Many distinct names used in the standard documentation and other literature to refer to the same thing.

Specification of parameters, their _matching_ with actual arguments, etc. feels like an unnecessarily complicated (although powerful) part of Python. This increases the time necessary to learn Python _properly_. To add to the complexity, certain syntax is only available in or after some minor versions (e.g. 3.8) of the language!

Formal parameters are a part of the function _signature_.

There are two types of parameter:

1) Positional. Roughly speaking, the _position_ of such arguments (in a list of parameters) matters. They are identified by a name. Function calls only provide values as positional arguments.
2) Named. Roughly speaking, the _name_ of such argument matters. They are specified usually as: `<name>=<expression>`. They are also known as _default_, _optional_, or _keyword_ parameters. Thus, every named parameter has a default value that is calculated exactly once, that is, when the `def` statement is interpreted (i.e. function code is compiled and the corresponding function object is created). Function calls provide both name and value as arguments for these.

##### Positional-only Marker
> A function signature may contain a single positional-only marker (/) as a dummy parameter. The parameters preceding the marker are known as _positional-only_ parameters, and must be provided as positional arguments, not named arguments, when calling the function; using named arguments for these parameters raises a `TypeError` exception.

##### Positional Argument Collector
This is really weird. I am going to pass it. May revisit later. At least right now (on a first reading) I think that this may not qualify to be one of _Python: The Good Parts_. This is also not very well explained in this book. Arguably, ChatGPT does a better job:

> In Python function signatures, the positional argument collector is the construct that **gathers (collects) any extra positional arguments passed to a function into a tuple**.

> It is written using `*` followed by a name, usually `*args`. PEP 3102 calls it "varargs".

> When you omit then name of a collector in the function signature, it becomes a _bare positional collector_ which does not collect anything, just blocks. Thus, the function `def f(*, a, b)` only accepts keyword, or named parameters, not positional parameters: `f(b = 5, a = 10)` and `f(a = 3, b = 7)` are valid calls, `f(10, 5)` isn't. 

````python
def f(*, a, b): # allows keyword parameters only
    print(f"a={a}, b={b}")
f(b = 5, a = 10)
f(a = 3, b = 7) 
f(10, 5) # not allowed
````

##### Named (Keyword) Argument Collector
Here again, the description in this book gives an impression that this _feature_ is erudite. I asked ChatGPT to summarize it for me and it gave me the following succinct table:

##### Summary of the Two Collectors
| Syntax     | Collector type       | Stores into | Meaning                        |
| ---------- | -------------------- | ----------- | ------------------------------ |
| `*args`    | positional collector | tuple       | collects extra positional args |
| `**kwargs` | keyword collector    | dict        | collects extra keyword args    |


The main application of the keyword argument collector is to provide callers the flexibility of providing optional arguments as `<name>=<value>` pairs which, inside the function body, can be updated in a named dictionary. This avoids telescoping (long) method signatures.

### The `return` Statement
You can use the `return` keyword in Python only inside a function body, and you can optionally follow it with an expression.

### Calling Functions
- Syntax: `<function-object>(<arguments>)`.
- When the function call executes, the parameter names are bound to the argument values in a *new namespace* (we'll study this soon), the function body executes, and the value of the function call expression is whatever the function returns. Objects created inside and returned by the function are liable to garbage collection unless the caller retains a reference to them.

#### Positional and Named Arguments
A Few examples are in order.

````python
def f(a, b, c=23, d=42, *x):
    print(a, b, c, d, x)

f(1,2) # pure positional, defaults apply
f(1,2,3,4,5,6) # positional, override defaults, capture remaining in x
f(1)
````

````python
f(a=1, b=2) # positional parameters can be passed using name
````

````python
def divide(divisor, dividend=94):
    return (dividend // divisor, dividend % divisor)

print(divide(30))
print(divide(divisor = 40))
print(divide(50, 100))
````

````python
a = 24
b = 95
c = 83
xs = [a, b, c]

def f(*numbers):
    print(type(numbers))
    return sum(numbers)
print(f(a, b))
print(f(*xs))
````

### Namespaces
A function's parameters and any names bound (by assignment or by other binding statements, such as `def`) in the function body make up the function's _local namespace_, also known as its _local scope_. Each of these variables is known as a _local variable_ of the function.

Variables not local are global. Global variables are attributes of the `module` object (we'll study them shortly).

In general, namespaces are programming languages' generalization of the concept of "context" in spoken languages.

#### Closure

> A function literal can appear anywhere that an expression can appear. Functions can be defined inside other functions. An inner function, of course, has access to its parameters and variables. An inner function also enjoys access to the parameters and variables of the functions it is nested within. **The function object created by a function literal contains a link to that outer context. #### Lambda Expressions
If a function body is a single return expression statement, you may (occasionally!) choose to replace the function with the special lambda expression form:

`lambda <parameters>:<expressions>`

A lambda expression is the _anonymous_ equivalent of a normal function whose body is a single `return` statement. The lambda syntax does not use the `return` keyword. You can use a lambda expression wherever you can use a function reference. Lambda can sometimes be handy when you want to use an elementary function as an argument or return value.

In functional languages, Lambda Expressions (or S-Expressions) are essential. In Python, however, they are syntactic sugar:

````python
xs = [-2, -1, 0, 1, 2]
sorted(xs, key=lambda x: x * x)  # returns: [0, -1, 1, -2, 2]
````

````python
# alternatively, use the good ol' def. They are easier to read, and readability counts.

def square(value):
    return value * value
sorted(xs, key=square)           # returns: [0, -1, 1, -2, 2]
````

### Generators or Generator Functions
The PEP on Generators is [PEP-255 on Simple Generators](https://peps.python.org/pep-0255/), which requires understanding [PEP-234 on Iterators](https://peps.python.org/pep-0234/). Whereas Iterators (PEP-234) sound straightforward now (modifying the behavior of the `for` loops in association with _Iterables_ in a syntactically pleasant manner), Generators are nuanced. Here's the motivation (from PEP-255):
> When a producer function has a hard enough job that it requires maintaining state between values produced, most programming languages offer no pleasant and efficient solution beyond adding a callback function to the producer's argument list, to be called with each value produced.

Consider writing a _producer_ function (a function that produces things), `tokenize`, that searches text and announces tokens of interest, preferably _as soon as_ one is found.

One view is to search the text and return the entire list of tokens of interest. That requires 1) the text is _not_ unending, 2) the caller is willing to wait for the function to return, and 3) some way to store the tokens found.

If any of these conditions can not be satisfied, then we have a programming language design problem. We may avail a callback mechanism in which the caller supplies a callback function (as an argument) to the producer (which the producer calls). This assumes that "code" can be passed as a value (functions as a first-class citizen).

Threads of execution can also be used. This requires _shared_ thread-safe data structures that multiple threads use: the producer adds to the structure and the consumer removes from it.

An attractive option for the producer is to _yield_ a value as it becomes available. Thus, the producer behaves like an Iterable. Keep in mind that this is a higher-level language feature; a lot goes on under the hood, which can be revealed by analyzing the bytecode. Here is a blurb of Generators from the book:
> When the body of a function contains one or more occurrences of the keyword `yield`, the function is known as a generator, or more precisely, a generator function. When you _call_ a generator, the function body does not execute. Instead, the generator function returns a special iterator object, known as a generator object (sometimes, quite confusingly, also called just 'a generator'), wrapping the function body, its local variables (including parameters), and the current point of execution (initially, the start of the function).

For a _`class`_ to create objects that are Iterable (i.e., targets in `for` loops), the `class` must implement an `iter` method. But with generator functions, it's even easier: you only need the `yield` expression in the function body. Then, the `for` loop becomes: `for <loop-variable> in <generator-function>(<arguments>)`. Note that `<generator-function>(<arguments>)` **calls the generator function that returns an object**. Then, an implicit (or explicit) call to `next()` results in returning the value `yielded` by the generator function body. Clearly, closures are required here because the `yield` expression must retain the state of the outer function, from which the control had `returned` before.

There's also a `yield from <iterator>` expression that conveniently `yields` elements. This is like _chaining_ the generator functions.

For most of the programming needs, `yield <expression>` and `yield from <generator>` are sufficient.


````python
def updown(N): # yields 1, 2, ...,N-1, N, N-1, ..., 1
    for x in range(1, N):
        yield x
    for x in range(N, 0, -1):
        yield x
for i in updown(3): # a call to the generator function returns a generator object at this call-site
    print(i)              # prints: 1 2 3 2 1
````

````python
# Equivalent code with `yield from`
def updown(N):
    yield from range(1, N)
    yield from range(N, 0, -1)
for i in updown(3):
    print(i)              # prints: 1 2 3 2 1
````

Advanced uses of generator functions include writing _coroutines_. 

Coroutines are an advanced topic, although they were introduced in 1958 by Mel Conway (according to Wikipedia). [This SO answer](https://stackoverflow.com/a/23436125/437506) tries to elucidate the difference between a thread and a coroutine:

> **Short answer**: With threads, the operating system switches running threads **preemptively** (according to the preference) according to its scheduler, which is an algorithm in the operating system kernel.
> (The dictionary meaning of preempt: take the place of or have precedence over)
> With coroutines, the programmer and programming language determine when to switch coroutines; in other words, tasks are cooperatively executed (cooperative multitasking) by pausing and resuming functions at set points, typically (but not necessarily) within a single thread.

> Long answer: In contrast to threads, which are preemptively scheduled by the operating system, coroutine switches are cooperative, meaning the programmer (and possibly the programming language and its runtime) controls when a switch will happen ...

[PEP-342 explains the motivation](https://peps.python.org/pep-0342/#motivation) behind coroutines and the support for writing them in Python:

> Coroutines are a natural way to express many algorithms, such as simulations, games, asynchronous I/O, and **other forms of event-driven programming or cooperative multitasking**. Python's generator functions are almost coroutines--but not quite--in that they allow pausing execution to produce a value, but do not provide for values or exceptions to be passed in when execution resumes. They also do not allow execution to be paused within the try portion of try/finally blocks, and therefore make it difficult for an aborted coroutine to clean up after itself.

> Also, generators cannot yield control while other functions are executing, unless those functions are themselves expressed as generators, and the outer generator is written to yield in response to values yielded by the inner generator. This complicates the implementation of even relatively simple use cases like asynchronous communications, because calling any functions either requires the generator to block (i.e., be unable to yield control), or else a lot of boilerplate looping code must be added around every needed function call.

> However, if it were possible to pass values or exceptions into a generator at the point where it was suspended, a simple co-routine scheduler or trampoline function would let coroutines call each other without blocking – a tremendous boon for asynchronous applications. Such applications could then write co-routines to do non-blocking socket I/O by yielding control to an I/O scheduler until data has been sent or becomes available. Meanwhile, code that performs the I/O would simply do something like this: `data = (yield nonblocking_read(my_socket, nbytes))
`
> in order to pause execution until the nonblocking_read() coroutine produced a value.

> In other words, with a few relatively minor enhancements to the language and to the implementation of the generator-iterator type, Python will be able to support performing asynchronous operations without needing to write the entire application as a series of callbacks, and without requiring the use of resource-intensive threads for programs that need hundreds or even thousands of cooperatively multitasking pseudo-threads. Thus, these enhancements will give standard Python many of the benefits of the Stackless Python fork, without requiring any significant modification to the CPython core or its APIs. In addition, these enhancements should be readily implementable by any Python implementation (such as Jython) that already supports generators.

This is advanced (and refers to a history of Python) because it assumes an understanding of a programming paradigm called Event-driven programming. I haven't personally done much event-driven programming.

#### Generator Use-cases (Especially As Coroutines)

##### Finding The Running Average

````python
def p():
    print("hello, world!")
print(type(p))

def running_average():
    """
    Remembers the state across invocations
    """
    total = 0
    count = 0
    avg = None
    total_s = ""
    ret_val = ""

    while True:
        x = yield ret_val     # yield current avg, wait for next number
        total += x
        count += 1
        total_s += f"{str(x)}" if x == 0 else f"+{str(x)}"
        avg = total / count
        ret_val = f"{total_s}/{count} = {avg}"

c = running_average() # receives a generator object
print(type(c)) # => <class 'generator'>, not <class 'function'>
next(c)                 # prime
for x in range(10):
    print(f"sent {x}, received running average so far {c.send(x)}")
````

##### Pipeline of Coroutines

````python
def grep_coroutine(pattern):
    print(f"Grep coroutine for pattern: {pattern}")
    try:
        while True:
            line = yield
            if pattern in line:
                print("MATCH:", line)
    except GeneratorExit:
        print("grep coroutine closing")

def producer(coro):
    coro.send(None)  # prime it, otherwise a TypeError: can't send non-None value to a just-started generator
    for line in ["apple", "banana", "pineapple", "orange"]:
        coro.send(line)
    coro.close() # systematically close the coroutine.

g = grep_coroutine("apple")
producer(g)
````
