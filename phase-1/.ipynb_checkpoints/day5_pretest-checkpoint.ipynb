### Expressions and Operators

- An expression produces a value.

Here's a list of operators in a decreasing order of _precedence_. Operator precedence is commonly understood. This is an aftermath of the prevalence of the infix notation. We can agree that infix notation (e.g `2 + 3`) is more readable than either prefix or postfix.  
- Dictionary creation: `{<key_expr>:<value_expr>, ...}`
- Set creation: `{<expr>, <expr>, ...}`
- List creation `[<expr>, <expr>, ...]`
- Tuple creation: `(<expr>, <expr>, ...)` -- note: `()` are recommended, not required.
- Function call: `<function_name>()` -- Associates Left -> Right -- note: white-space allowed, but not recommended between the `<function_name>` and `(`.
- Slicing: `<iterable_name>[<index_expr>:]` or `<name>[<index_expr>:<step_expr>]` -- Associates Left -> Right
- Indexing: `<iterable_name>[<index_expr>]` -- Associates Left -> Right
- Attribute reference: `<reference_name>.<attr_name>` -- note: evaluates to the value of the attribute if it exists, can be a callable (function) or not -- Associates Left -> Right 
- Exponentiation or Power operation: `<base_expr>**<exponent_expr>` -- note: `<base_expr>` to the power `<exponent_expr>` -- Associates **Right -> Left**
- Bitwise NOT and unary operations: `~<number>`, `+<number>`, `-<number>` -- note: **curiously**, they can be chained: `x=--2 # value of x is 2`
- Multiplication: `<multipliable-number> * <number>` -- note: `"a" * 3` yields `"aaa"`. Matrix Multiplication: `<matmul_expr> @ <matmul_expr>` -- note: Due to a Python Enhancement Proposal (PEP) specifically for the `numpy` library which is useful for matrix multiplication; not available in standard Python, although grammar allows that for the more readable syntax of the code that uses `numpy` which yields multiplication of matrices `x @ y` if `x`, `y` are matrices. Division: `<number_expr> / <number_exp` -- note: Yields the quotient that the _true division_ of dividend and divisor will return. **Floor Division**: `<dividend_expr> // <divisor_expr>` -- note: Specific to Python; it yields the _mathematical floor_ of the quotient; closely related to the `%` operator. Remainder: `<dividend_expr> % <divisor_expr>` -- note: There are many ways to remember what Python does in this case; I remember 1) The division algorithm, and 2) Python gives the same sign to the remainder that the divisor has. **Curiously, `%` can be applied to floats too!** -- perhaps it has some application. But who has used it? (All of these associate Left -> Right)

````python
7 % 3 # 1, straightforward application of Euclidean division: a = b × q + r
````

````python
-7 % 3 # -7 = 3 × -2 + (-1) or 3 × -3 + 2, but since the remainder should have the sign of divisor, this should be 2
````

````python
7 % -3 # 7 = -3 × -2 + 1 or -3 × -3 + (-2), but since the remainder should have the sign of the divisor, this should be -2
````

````python
-7 % -3 # -7 = -3 × 2 + (-1) or -3 × 3 + 2, but since the remainder should have the sign of the divisor, this should be -1
````

- Note: The builtin function `divmod` agrees with the floor division and remainder operations.

````python
def does_divmod_agree_with_floor_div_remainder(a, b):
    qr = divmod(a, b)
    print("floor division agrees?: ", qr[0] == a // b)
    print("remainder agrees?: ", qr[1] == a % b)

t = does_divmod_agree_with_floor_div_remainder # interesting little application of the fact that functions are values!
t(7, 3)
t(7, -3)
t(-7, 3)
t(-7, -3)
````

- Addition: `<number_expr> + <number_expr>`, Subtraction: `<number_expr> - <number_expr>` -- note: Associates L -> R
- Left shift, Right shift: `<int_expr> << <int_expr>>`, `<int_expr> >> <int_expr>>` -- note: Associates L -> R
- Bitwise AND, XOR, OR: `<int_expr> & <int_expr>>`, `<int_expr> ^ <int_expr>>`, `<int_expr> | <int_expr>>` -- note: Associates L -> R
- Comparison Operators: `<`, `<=`, `>`, `>=`, `!=`, `==` -- note: can be chained -- `x < y < z`
- Identity tests (determine if values refer to the same object or not): `<expr> is <expr>`, `<expr> is not <expr>`
- Membership tests: `<container_expr> in <value_expr>`, `<container_expr> not in <value_expr>`
- Boolean NOT: `not <expr>` : note: a welcome change, there is no `&&`. However, `<expr>` **need not be an expression that evaluates to a boolean value**; in this regard, Python is like C. Associates L -> R
- Boolean AND: `<expr> and <expr>` -- note: Associates L -> R
- Boolean OR: `<expr> or <expr>` -- note: Associates L -> R
- Conditional expression (ternary operator): `<expr_1> if <c_1> else <expr_2>`
- Anonymous simple function: `lambda [<param>, <param>, ...] : <expr>`
- Assignment expression(!) : `<identifier>:=<expr>`

- In the conditional expression (ternary operator): Only one of the expressions `when_true` and `when_false` is evaluated.

````python
import random

def when_true():
    print("when_true")
    return 1
def when_false():
    print("when_false")
    return 0
condition = random.choice([True, False])
# exactly one of when_true and when_false is called and the corresponding value is returned
print(when_true() if random.choice([True, False]) else when_false())
````

#### Assignment Expressions Using `:=`
- This is an interesting feature that combines evaluation and assignment inside a (control) statement! This shows that programming languages evolve and Python has a process to do that. It was added in Python 3.8 (by incorporating PEP-572). It enhances programming experience of a few common programming patterns in imperative programming. It aligns well with the dynamic typing of the language. Here is a compelling excerpt from that [proposal](https://peps.python.org/pep-0572/#the-importance-of-real-code):

> During the development of this PEP many people (supporters and critics both) have had a tendency to focus on toy examples on the one hand, and on overly complex examples on the other. The danger of toy examples is twofold: they are often too abstract to make anyone go "ooh, that's compelling, and they are easily refuted with "I would never write it that way anyway". The danger of overly complex examples is that they provide a convenient strawman for critics of the proposal to shoot down ("that's obfuscated"). Yet there is some use for both extremely simple and extremely complex examples: they are helpful to clarify the intended semantics. Therefore, there will be some of each below.

> However, in order to be compelling, examples should be rooted in real code, i.e. code that was written without any thought of this PEP, as part of a useful application, however large or small. **Tim Peters has been extremely helpful by going over his own personal code repository and picking examples of code he had written that (in his view) would have been clearer if rewritten with (sparing) use of assignment expressions. His conclusion: the current proposal would have allowed a modest but clear improvement in quite a few bits of code.**

The entire proposal is highly readable.

Luckily, it also does not clutter the source code line; subtly adds a single character `:`. However, it does make you improve your understanding of how _assignment_ is handled in Python (as a statement, an expression, or both).

##### 1. `:=` in `if` or `elif` Statements
```
# 1) common pattern
flag = <expression> # standalone evaluation and assignment
if flag:
  ...
# 2) collapse in single statement inside the if
if flag := <expression>:
  ...
```
Who wouldn't prefer 2) to 1)?

##### 2. `:=` in the `while` Statement with a Filter
- What is best way to write code that iterates over a collection, filters out any unwanted elements, and processes the remaining? If you say, "Use an Iterator", then perhaps you don't need this feature. But if you want to write a traditional `for` or `while` loop, you might.

##### 3. `:=` in a List Comprehension
We'll revisit this.


If `:=` is so good, why not use it everywhere? In other words, why not replace every use of `=`, which is assignment _statement_, with that of `:=`, which is assignment _expression_? PEP 572's [clarification](https://peps.python.org/pep-0572/#why-not-just-turn-existing-assignment-into-an-expression):

> C and its derivatives define the `=` operator as an expression, rather than a statement as is Python’s way. *This allows assignments in more contexts, including contexts where comparisons are more common* (in those languages). The syntactic similarity between `if (x == y)` and `if (x = y)` belies [^1] their drastically different semantics. **Thus this proposal uses `:=` to clarify the distinction.**

^1: contradicts, misrepresents

It's in programmers' interest that this is clear. Although Java derives its syntax from C, because of its philosophy, (compile-time) type-safety is taken seriously. In C, since there is no separate boolean type, we often see and accept (again, get used to) code like this:
```
int flag;
if (flag = get_flag()) { 
    /* 
      note '=', since it is an expression in C, it is allowed, and even acceptable, in more contexts like this one!, 
      also since if (0) is analogous to if (false) this style is preferable to:
      if ((flag = get_flag()) != 0) 
         ...
    */
}
```
In Java, this changes little even because of the focus on type-safety (you declare the type of `flag` as `boolean`):
```
boolean flag; // has a default value false
if (flag = getFlag1()) { //getFlag1() must return a boolean or Boolean
    // ...
} else if (flag = getFlag2()) {
    // ...
}
```
or, if `flag` is an object:
```
Flag flag; // has a default value null
if ((flag = getFlag1()) != null) { // getFlag1() must return an object reference
    // ...
}
else if ((flag = getFlag2() != null) {
    // ...
}

```

But in both the languages (not to mention C++), *since `=` operator creates an expression*, it is allowed (and acceptable) in more contexts. Therefore, it is a common practice to use it inside `if`.

Not in Python. In Python `=` is a statement, not an expression. If Python compiler encounters code like:
```
if f = get_flag():
    #... use f here
```
then it flags an error: `SyntaxError: invalid syntax. Maybe you meant '==' or ':=' instead of '='?`

Thus, `:=` clarifies the programmer's intent: assignment immediately followed by comparison. It is even more useful when you want to use the object assigned later.
