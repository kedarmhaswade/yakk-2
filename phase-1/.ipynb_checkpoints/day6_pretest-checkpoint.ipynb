### Numeric Operations
#### Intro
- Nothing too surprising here. We have seen `+,-,/, //, %` above. There could be some surprises with respect to the types of operands and results of numerical operations. This too, aligns well with the dynamic nature of Python.
- Python has a builtin complex number type to deal with complex numbers.
- Numbers are immutable. This is a good thing.


#### Numeric Conversions
- For _binary_ operators, if the operand types differ, Python _widens_ the _narrower operand_. Of the three numeric types, `int` is the narrowest, followed by `float` and then `complex`. The conversion of `complex` number to `int` or `float` isn't allowed.

### Sequence Operations
#### Intro
- Think of "containers", "sequences", and "iterables". They are carefully distinguished.
- Standard sequences: Strings, Lists, Tuples. Sequences are _ordered containers_ with items that are accessible by _indexing_ and _slicing_.
- Containers: All standard sequences and Sets and Dictionaries.
- Iterables: "Objects over which you can loop" -- these are all containers and more, like files and _generators_.

#### General Sequence Operations
- `len(obj)`: Returns number of items in the container.
- Overloaded methods `max` and `min` return the maximum and minimum of _totally-ordered, comparable elements_ in various flavors. An argument named `key` they take is called the _keyword-only_ argument (we'll learn about this soon) which denotes the name of the "comparator function".
- If you have a list, just pass it to the `max` or `min` functions and they'll do the expected thing. It can't get better than this. Try writing code to do the same thing in Java (even after Java 8)--it takes quite a few keystrokes. Python is truly a VHLL (Very High Level Language) in this regard.
- There's a `sum` function that sums the numbers in an iterable and it is _intended specifically for use with iterables of numeric values and may reject non-numeric types_.

#### Sequence Conversions
- No implicit conversion between different sequence types.
- Functions `tuple(<iterable>)` and `list(<iterable>)` return a new tuple and a new list of items from the given iterable.

#### Concatenation and Repetition
- `+` is overloaded to do the expected thing (concatenation) with sequences, strings, tuples, and lists.
- `*` : `s*n` is the concatenation of `n` copies of the sequence `s`.

#### Membership Testing
- `<x> in <xs>` tests whether the sequence has any `object` that _equals_ (will revisit object equality) `<x>`.
- `<k> in <dict>` tests whether there is a mapping in `<dict>` for the _key_ `<k>`.
- `<s> in <str>` tests whether the string `<str>` has a _substring_ `<s>`.

#### Indexing a Sequence
- Indexing is 0-based.
- `<seq>[n]` is the `n`th element.
- If `<seq>` contains `L` elements, and `i` is a positive integer less than or equal to `L`, then `<seq>[-i]` and `seq[L-i]` refer to the same element.

```python
xs = [10, 22, 453, -24]
L = len(xs)
for i in range(1, L + 1):
    print(xs[-i] == xs[L - i])
```

#### Slicing a Sequence
- This is a great way to do many things (including taking subsequences) from a sequence. It is also a good example of how to design an API.
- Practice with it to improve.
- The general syntax: `<seq>[<start-index>[:<stop-index>][:<stride>]]`. Observe defaults.
- Nothing surprising here. This is everything you'd expect from a VHLL.

#### Tuples
- Tuples are immutable; you can't rebind (reassign) or delete (`del`) an item or slice of it. That doesn't mean Python can enforce immutability of its elements themselves. Java, because of its focus on type-safety, can. One can declare an "immutable list of immutable objects" in Java and expect it to behave accordingly. Not in Python. That is not a limitation, it's a feature: Compile-time type-safety is not a core design principle of Python, whereas it is an enforced afterthought in Java (after Java 5, but that was quite a while ago; for all practical purposes, it is a core design principle of Java).
- Of course, nobody can prevent you from creating a tuple of immutable objects, but a tuple can only resist _structural_ modifications.
- Two normal methods available to tuples: `count()` and `index`

```python
t = (5, -3, 22)
t.count(0) # counts the number of times 0 has come in t, returns 0
t.index(4) # unfortunately, this throws an error: ValueError: tuple.index(x): x not in tuple; would you have preferred returning an index of -1?
# no, because Python supports negative indices!
t[t.index(-3)] # unreachable, comment above line and rerun this cell; => -3
```
