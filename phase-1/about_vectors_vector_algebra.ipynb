## 1. Components

Given a vector $\vec{V}$ and two non-coincidental lines, AA' and BB', we can always find vectors $\vec{X}$ and $\vec{Y}$ _uniquely_ along those two lines such that $\vec{V}=\vec{X}+\vec{Y}$ (the vector addition and the parallelogram law.)

See Figure 1.1.



## Exercise 1.1 
Prove the uniqueness of $\vec{X}$ and $\vec{Y}$.

### My Solution 1.1
Geometry of parallelogram is clear. Consider two parallelograms, OSRT and OS'RT' where OS and OS' are along the same direction and OR and OR' are also along the same direction (other than OS). If S' does coincide with S, then OS'RT' is not a parallelogram because S'R is not parallel to OT'. 

---

Let $\vec{e_x}$ and $\vec{e_y}$ be nonzero vectors starting at O and lying along $\vec{X}$ and $\vec{Y}$ respectively, such that 

$\begin{align}
\vec{X}&=V_x\vec{e_x} \\
\vec{Y}&=V_y\vec{e_y}
\tag{1.1}
\end{align}$

(by simple _scaling_ of vectors $\vec{e_x}$ and $\vec{e_y}$).

- $V_x$, $V_y$, which are scalars, are called the _components of $\vec{V}$ relative to or in the _frame of reference_ defined by $\vec{e_x}$ and $\vec{e_y}$. Consequently, $\vec{X}$, $\vec{Y}$ are _component vectors_ of $\vec{V}$.
- Point O is called the _origin of the frame of reference_.
- The vectors $\vec{e_x}$, $\vec{e_y}$, whose magnitude is (typically) 1, are called the _base vectors_. (I have also heard the term _unit vector_.) 

## Exercise 1.4
If $\vec{e_x}$ and $\vec{e_y}$ have unit magnitude and are perpendicular, what are the components of a vector of magnitude $h$ _pointing along the positive direction of_ $\vec{e_x}$? Would the components be different if $\vec{e_y}$ were not perpendicular to $\vec{e_x}$? 

### My Solution 1.4
If the vector points in the direction of $\vec{e_x}$, the only parallelogram is the one whose two sides are of length zero and two of length (corresponding to) $h$: The parallelogram reduces to a line. Regardless of whether $\vec{e_x}$ and $\vec{e_y}$ are perpendicular, the components must be $(h, 0)$.
