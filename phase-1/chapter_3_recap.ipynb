      - Consider finding the largest element in a list, or the sum of list elements, as you are populating it. My instinct is to do it serially by maintaining state variables. The Pythonic way seems to be doing it after the fact:

``` {python}
cart = [(name, price)]
# loop to populate the cart: cart.append(("book", 10.3)), 
# do not maintain the sum, most expensive item, etc.
total = sum(price for _, price in cart) # the argument is a "generator expression"
max = max(cart, key = lambda name_price: name_price[1])
avg = total / len(cart)
```
      
- Is that less efficient? For really long collections, it is. However, a shopping cart is not expected to have more than a few dozen items. For such small collections, readability trumps half-baked efficiency concerns. For large collections, we need to use specialized libraries optimized for heavy-duty processing. Python has excellent libraries for data processing. **Python excels at expressive aggregations**.

## Comparison of dict and `collections.defaultdict`

|  | dict | `defaultdict` |
| :------- | :------ |:-------|
| Missing key access|`KeyError` (can be worked around with `setdefault`)|Auto-create with default factory|
| Typical use | General Mapping, small databasets|Grouping, counting, multimaps, production datasets|
|Performance in grouping tasks|OK|Faster, cleaner|


````python
# Counting 
# with `defaultdict`:
from collections import defaultdict
data = [1, 2, 2, 3, 3, 3]
counts = defaultdict(int)

for x in data:
    counts[x] += 1
print(counts)
# with the builtin
counts = {}
for x in data:
    if x in counts:
        counts[x] += 1
    else:
        counts[x] = 1
print(counts)
````
