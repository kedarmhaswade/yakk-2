## Classes and Instances
A class is a user-defined type, which you instantiate to build instances, i.e., objects of that type. Python supports this through its class and instance objects.

### Python Classes

A class is a Python object. Here are its characteristics:

- You can call a class object just like you'd call a function. Thus, a class is a _`callable`_.
- A class has arbitrarily named attributes that you can bind to values that we call _descriptors_. You can also use these names to reference the descriptors.
- When a descriptor is a function, it's called a _method_.
- Special methods, named dunder methods, embody Python's "duck-typing".
- Classes can _inherit_ from other classes.

#### ❓ Mini Quiz

Guess the output of the call of `tricky`:

```
def tricky(a, b=1, /, c=2, *args, d=3, **kw):
    print(a, b, c, args, d, kw)

tricky(10, 20, 30, 40, 50, d=99, e=100)

```

<details>
<summary><strong>Answer</strong></summary>

Binding sequence

1) Positional-only
   - `a = 10`
   - `b = 20`

2) Positional-or-keyword
   - `c = 30`

3) Extra positional
   - `args = (40, 50)`

4) Keyword-only
   - d = 99

5) Extra keyword
   - `kw = {'e': 100}`

Output:

`10 20 30 (40, 50) 99 {'e': 100}`

Remember:
- `/` restricts how earlier parameters may be passed (positionally or as keywords); `/` is the positional marker.
- `*` restricts how later parameters must be passed (only as keywords). `*args` is the positional argument _collector_.
