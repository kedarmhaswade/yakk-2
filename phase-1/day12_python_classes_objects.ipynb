## Classes and Instances
A class is a user-defined type, which you instantiate to build instances, i.e., objects of that type. Python supports this through its class and instance objects.

### Python Classes

A class is a Python object. Here are its characteristics:

- You can call a class object just like you'd call a function. Thus, a class is a _`callable`_.
- A class has arbitrarily named attributes that you can bind to values that we call _descriptors_. You can also use these names to reference the descriptors.
- When a descriptor is a function, it's called a _method_.
- Special methods, named dunder methods, embody Python's "duck-typing".
- Classes can _inherit_ from other classes.

Remember:
- `/` restricts how earlier parameters may be passed (positionally or as keywords); `/` is the positional marker.
- `*` restricts how later parameters must be passed (only as keywords). `*args` is the positional argument _collector_.
- `/` (which guards the left side) **prevents** callers from relying on parameter names. 
- `*` (which guards the right side) **prevents** callers from relying on parameter position.
- The three 'regions' of a Python signature: `def f(pos_only, /, pos_or_kw, *, kw_only)`. Simply speaking, if either marker is missing, that region doesn't exist.
- A _bare_ `*` ends the positional arguments, `*args*` permits and captures (absorbs) them. Any "following" arguments must be named. `def mean(a, b, *, weights=none)` _forces_ you to pass exactly two values (whose arithmetic mean is to be calculated), whereas `def mean(a, b, *more, weights=none)` absorbs additional values (presumably to calculate the arithmetic mean of all the values) in a tuple named `more`.

Every class extends the class `object`. `<statements` comprise the _class body_ which must finish executing before the new class object named `<ClassName>` (in title case) is created.

Every class in Python has a _metaclass_, which we'll examine later. _Inheritance_ is a central concept in OOP. We'll see that later. For now, know that a class can _extend_ other classes. 

We saw earlier that when the Python interpreter encounters a `def` statement, it only compiles it to bytecode (in the form of an in-memory code object), creates a corresponding function object, and binds the function name to it. By contrast, when it encounters a `class` statement, the interpreter: 
- Compiles the class body to a code object
- Creates a new namespace (dict-like)
- Executes the class body immediately in that namespace
- Calls the metaclass (usually `type`) with:
- Binds the resulting class object to the class name.

The class body _is executed_ when the interpreter encounters a class definition. Note, however, that although the class body is executed, the function bodies inside it are not.

### The Class Body
The body of a Python class feels strange to someone familiar with a more structured format like a Java class. This experience is arguably made worse (at least at first sight) by naming conventions. Some of this is to be expected because Python supports duck-typing much more implicitly than Java.

### Functions Definitions in a Class Body (Methods)
Functions defined inside class body (methods) are important attributes for most class instances.

We must define an instance of a class. Generally speaking, an instance is the holder of the data and class holds the code. An  instance of a class is independent from every other instance, although they share the same code.

A method definition (the `def` statement appearing in a class body) receives a mandatory first parameter, conventionally named `self` that *refers to the instance* on which the caller calls the method (using the `.` operator).

````python
class C1:
    def hello(self):
        print(f"'hello' called on {id(self)}")

c1 = C1()
c1.hello()
c2 = C1()
c2.hello()
````

## Instances
Since a class is _callable_, we can create its _instances_. e.g. `c = C()` _instantiates_ a class whose name is `C` and binds it to a name `c`. We also say that `c`'s _type_ is `C`.

### `__init__`
Other languages call this special method constructor. If defined, it is called whenever a class is called. It is a hook provided for the class to do any initialization tasks. It mustn't return any value other than `None`. This is usually achieved by a missing `return` statement.

````python
class C6:
    def __init__(self, n):
        self.x = n

c6 = C6(200)
print(c6.x) #=> 200
````

### Attributes of Instance Objects
Once you have created an instance, you can access its attributes (data and methods) using the dot (`.`) operator.

There's some obscurity when it comes to data encapsulation in Python by default. Consider:

````python
class C7:
    pass
z = C7()
z.x = 23
print(z.x) # prints 23
````

The above code creates an _instance variable_ bound to the name `x` inside the class `C7` and assigns `23` to it. This could have been better achieved via the `__init__` method.

### Factory Function Idiom
Programming is heavily influenced by (common) patterns of design and implementation at various levels of sophistication. At a preliminary level of object-oriented programming, one may want to create instances of different classes based on certain parameters.

Creation of artificial (man-made) things usually happens in factories (since the time of industrial revolution), such functions that manufacture instances on demand are called factory functions.
