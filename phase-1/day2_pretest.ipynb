If your representation rescales magnitudes by a function s(m) (here $𝑠(𝑚)=𝑚^2$), then for two vectors pointing the same way you would need $s(a+b)=s(a)+s(b)\forall 𝑎,𝑏\geq 0$. But $s(m)=m^2$ does not satisfy additivity, so it cannot preserve vector addition or the parallelogram law in general. In short: any correct geometric representation that makes the parallelogram law hold for all vectors must scale lengths linearly with the vector magnitudes. *Squaring destroys that linearity.*

### 7. Some Awkward Questions
Using crisp arguments, Hoffmann asks a simple but complicated question, "When are two vectors equal?" (He refers to their representations.) In particular, any two arrows starting at the same point and having the same magnitude, direction, such that the law of parallelogram holds for their combination represent equal bound vectors in space. Here's Hoffmann, in his impeccable style:

Choosing a meaning for the word “equal” here is not as easy as one might imagine. For example, we could reasonably say that *two vectors having the same magnitudes, identical directions, and a common starting point are equal vectors*. And if one of the vectors were somehow pink and the other green, we would probably be inclined to ignore the colors and say that the vectors were still equal. **But what if one of the vectors represented a force and the other a displacement?** There would then be two difficulties.

Thus, there is something like a kind, type, or class that we intuitive take for granted. We know that we can't combine a force and a displacement, even if the law of parallelogram seems to hold for their representations. For numbers (or scalars), the _abstract idea_ of a number supersedes the concrete forms. For example, 3 apples and 2 oranges are neither 5 apples nor 5 oranges, but 5 _things_. The abstract idea of the number 5 still applies to the abstract idea common to both apples and oranges -- they are _thing_ s. With mere numbers, there are some other complications: Can 3 ft. combine with 3 in. to give 6 _things_, or do they combine to give 39 in. or $3\frac{1}{4}$ ft.?

There is no "vector equivalent" of combination of _countable numbers_. Apples are apples and Oranges are oranges and both are Things or Items. Force and Displacement are Vectors, and that's it. 

In object-oriented programming, we learn about _classes_ and _objects_. Classes may have _superclasses_. Thus the superclass of _Apple_ and _Orange_ is _Item_. Thinking of Apples and Oranges as Items lets us forget their differences and combine them as if all of them were either Apples or Oranges. What is the superclass of _Force_ and _Displacement_ that thus preserves their combination characteristics?

He uses a term "Olympian Simplicity". What does it mean?

Here's more of pure Hoffman:

We might seek to avoid trouble by retreating to the definition of a vector as "an entity having both magnitude and direction," _without_ mentioning the parallelogram law. But once we start retreating, where do we stop? 

Why not be content to define a vector as "an _entity_ having direction," or as "an entity having magnitude," or, with Olympian simplicity, as just "an entity”? 

Alternatively, we could make the important distinction between the abstract mathematical concept of a vector and entities, such as forces, that behave like these abstract vectors and are called vector quantities.

This helps, but it does not solve the present problem so much as sweep it  under the rug. We might amend our definition of a vector by saying that vectors combine according to the parallelogram law only with vectors of the same _kind_: forces with forces, displacements with displacements, accelerations (which are vectors) with accelerations, and so on. But even that is tricky since, for example, in dynamics we learn that force equals mass times acceleration. **So we would have to allow for the fact that though a force does not combine with an acceleration, it does combine with a vector of the type mass times acceleration in dynamics.**

We shall return to this matter. (See Section 8 of Chapter 2.) But enough of such questions here. _If we continue to fuss with the definition we shall never get started_. Even if we succeeded in patching up the definition to meet this particular emegency, other emergencies would arise later. 

**The best thing to do is to keep an open mind and learn to live with a flexible situation, and even to relish it as something akin to the true habitat of the best research.**
