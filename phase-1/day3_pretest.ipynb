# Day 3 (02 November 2025)
Learning a new programming language well (even when you are or have been a professional programmer) is difficult. Peter Norvig has famously written about it here: https://www.norvig.com/21-days.html. The essay is worth a read and I reread it. Of course, Norvig is a [beast of a programmer](https://news.ycombinator.com/item?id=15787023). He is a noted researcher and programmer. I must learn from him, but I should also remain grounded in my priorities by periodically asking, "Why am I here?", that is, "Why am I trying to become more comfortable with Python?".

I am comfortable with the Java way of writing a program (or large programs). This includes, to a good extent, using Java as it was meant to be used. I am somewhat familiar with SICP and functional abstractions. I wish I learn _Scheme_ someday.

I am trying to get more comfortable with Python because that's the tool of choice for AI or ML libraries. I want to get better at AI as (I think) it is practiced today in academia and industry. But that perception may be an illusion. However, my theoretical understanding of AI must be adequately supported by practical projects, a lot of which might have to do with data analysis for which Python ecosystem appears the most suited.

Python has been hailed [^1] has a good first programming language. However, it's not so for me. And yet, I don't want to adopt a "Learn in 21 days" approach. I do want to be guided by masters like Norvig and Guido. I want to keep my pursuit of understanding AI from the so-called [over-hype](https://www.odbms.org/blog/2025/10/beyond-the-ai-hype-guido-van-rossum-on-pythons-philosophy-simplicity-and-the-future-of-programming/), and for that, apart from understanding parts of its underpinning, learning to use Python libraries is required.

I believe the masters when they say that Python has certain elegance and simplicity. But Java masters (e.g. Josh Bloch) have said the same thing about Java and Oderski about Scala. Doug Crockford says the same thing about JavaScript (well, he primarily says JS functions are the best part of JS). Everyone loves Lisp and Scheme. A Rust cult is fast-developing for safe programming closer to the machine.

Revered gurus enlighten, but, I lament, they also tend to confuse. They are sitting on their high altars, having gotten their through (alleged) innate brilliance and deliberate practice, and dispensing good, experiential, time-tested, opinionated advice. I want to be guided by them, for sure. But am I so dedicated to a guru or a doctrine that I want to give up everything and spend the next 5 years just living that philosophy right from the word go? Gurus are also emotional humans. Even Guido, who calls AI a hype, now works for Microsoft, apparently, after it has been _expurgated_ of Steve Ballmer [^2].


Therefore, I must be self-aware. I am learning Python as a tool (and this doesn't mean I am relegating it to something inferior) to improve my understanding of AI. If Karpathy teaches how to implement Deep Learning using a few lines of Python, I want to learn from him without altering the tool. Of course, I can alter the tool and implement it in some other language, but why is that needed? There's already so much to understand! 

Sure, this is a view inspired by Economics (which helps us optimize resources to achieve a certain goal), but I am fine with that. 

I should correct the course all the time. Whenever I start to become comfortable with something, I should challenge myself by doing more difficult (but enjoyable) things. Starting small and keeping the journey arduous (but exciting) all the time seems like a good idea for a mere mortal like me. I am a present hedonist anyway. 

Why learn Python for 'AI'? That's a good question and it haunts me. Why not learn a few programming languages just to keep scratching the programming itch? Why not pick a book like CLRS, Bentley's Programming Pearls, Laaksonen's Competitive Programming, Charles Wetherell's amazing Etudes for Programmers (and Norvig's [Python reincarnation](https://github.com/norvig/pytudes) of it), Knuth et.al.'s Concrete Mathematics, Arthur Engel's or the like, and solve its problems as programs in a language you are learning?

That's a fine proposition and it is for life. There's no dearth of problems like this

If one wants to use a programming language that embodies principles of fault tolerance, one should give Erlang a serious consideration. Similarly, if one wants to understand AI for practical projects, Python has become a programming language of choice, even if its creator hates AI. He is not wrong, some companies want to project AI as a superpower driven by their business interests, but my interests are different. **I want to understand what makes current efforts and underpinnings of AI work[^3].** 

I could be a purist like the great mathematician Geoffrey Hardy, who advocated a math pursuit for the sake of it alone (and a part of me wants to do it), but that is a position I can take a little later in my journey. Hardy espoused his approach in his seminal (and controversial) book, _A Mathematician's Apology_, in which he famously claimed he had done nothing _useful_ in life (other than, of course, _discovering_ the genius of Ramanujan). Hardy is being too modest, of course. His great books (A Course in Pure Mathematics) and discoveries are a testament to his contributions.

I considered starting with the Python language reference. I also considered referring to Mark Lutz's mammoth Learning Python (now in 6th edition) and . I considered [John Zelle's Python Programming](https://mcsp.wartburg.edu/zelle/python/ppics4/index.html) now in 4th edition, based on Norvig's suggestion. Zelle's book appears attractive as a beginning programmer's book. I consider myself a graduating programmer who is perhaps a bit more experienced than an apprentice (although still a journeyman). 

Then I considered four more: 1) Python in a Nutshell by Martelli et. al. 2) Fluent Python by Ramalho 3) Python for Data Analysis by Wes McKinney, and 4) Deep Learning with Python, by Matthew Watson and Francois Chollet. 

Martelli's book appears like a good start. I will follow up with Fluent Python (which is like Josh Bloch's Effective Java) and then read the other two.

That's it. I am committing to it now. Next few days on Python in a Nutshell and Vectors by Banesh Hoffmann.


It's a lament of a practical Python practitioner (I do have a day job atm and I am burning the proverbial midnight oil to learn Python well) and also a decision in the hope of being self-aware.

Again, I am in no rush, but I want to make progress. Abstruse Goose (it's gone, but Internet Archive is there!) and Peter Norvig keep reminding me ...
