### Dictionary Operations
Dictionaries are quite useful. They are Python's implementation of a hashed structure. Indexing is done differently than than in a list. In other programming languages this structure is known as hash-map or associative array. In Computer Science, it is known as hash-table. It has been thoroughly studied. 

Simply speaking, a dictionary stores mappings of keys to values. Keys are expected to be immutable objects.

- The global function `len()` returns the number of mappings (key-value pairs or _kv_ pairs) it holds.
- A dictionary is an `iterable`. So, you can pass it to any function (`builtins.min`, `builtins.max`, etc.) that accepts an `iterable`. In this case, dictionary yields its keys in the insertion order. 

````python
d = {'a': 97, 'b': 98, 'c': 99}
print(len(d))
print(type(d))
````

#### Membership
- Same as that for a set; use the `in` operator: 

````python
d = {'a': 97, 'b': 98, 'c': 99, 'd': 100}
print('a' in d)
print('e' in d)
print('a' not in d)
print('e' not in d)
````

#### Non Mutating Methods
- `d.copy()`: Returns a _shallow copy_. Equivalent to `dict(d)`.

````python
name1 = 'Doe John'
address1 = {'street': '123 Fourth Ave', 'zip': '56789'} # value is a mutable dictionary

people = {name1:address1}

clones = people.copy() # shallow copy

print(people[name1] is clones[name1]) # they are the same object

clones2 = dict(people)
print(people[name1] is clones2[name1]) # they are the same object
````

- `d.get(k)` is a silent equivalent of `d[k]` under normal circumstances; the former returns `None` for non-existent keys whereas the latter raises a `KeyError` exception.

````python
print(d.get('e'))
d['e']
````

- Each of `d.items()`, `d.keys()`, `d.values()` returns an `iterable` view of key-value pairs, keys, and values in `d` respectively.

#### Mutating Methods
- `d.clear()`: It's meaning is clear.
- `d.pop(k, default)`: Removes and returns `d[k]` when `k` is `in` `d`, otherwise returns the `default` if provided, otherwise raises `KeyError`.
- `d.popitem()`: Removes and returns the item inserted last in `d`. The returned value is a 2-tuple: `(key, value)`
- `d.setdefault(k, x)`: Sets the default for the key `k` *and* returns the value. This method mutates the dictionary (unlike the `get` method that takes the default value). This method is (more) useful when the value is a collection. 

````python
d = {'a': 97, 'b': 98, 'c': 99, 'd': 100}
d.setdefault('a', 90)
print(d['a'])
print(f"popped: {d.pop('a')}")
d.setdefault('a', 90)
print(f"setdefault: {d['a']}")
print(f"popped: {d.pop('a')}")
````

- Because `setdefault` returns a value (typically "setters" do not return anything), you may be surprised. But that is helpful at times:

````python
# Consider this: A Person has a 'fleet' of pets. When they purchase a new pet, it's added to this fleet.
person = {'name': 'Joe'}
pk = 'pets'
# A dog was purchased
person.setdefault(pk, []).append('dog') # modifies the value because we get a reference to it which is a container (list)
print(person[pk])
del person[pk]
# without setdefault, we would need more code, something like the following or a variation thereof.
if not pk in person:
    person[pk] = []
person[pk].append('dog')
print(person[pk])
````

- It's difficult to escape the feeling that Python has been enhanced to provide attractive syntax to common programming situations (patterns).
- Even so, `setdefault` is not used much in modern Python. We will soon learn `defaultdict`from the `collections` module that modern Python uses.

- `d.update(d1)`: For each mapping in `d1`, sets `d[k] = d1[k]`. Same as `d |= d1`.

- The `items`, `keys`, and `values` methods return values known as _view objects_. If the underlying `dict` changes, the retrieved view also changes; Python doesn't allow you to alter the set of keys in the underlying `dict` while using a `for` loop on any of its view objects.

````python
d = {'a': 97, 'b': 98, 'c': 99, 'd': 100}
kview = d.keys()
for k in kview:
    print(f"{k}:{d[k]}")
d.popitem() # dictionary changed after iteration
for k in kview: # the view reflects it
    print(f"{k}:{d[k]}")
for k in kview:
    if k == 'a':
        d.pop(k) # RuntimeError: structural modification while iterating on a view
print(d)
````

- `d.pop(key, None)` is a useful shortcut for removing a key from a `dict` without having to first check if the key is present, much like `s.discard(x)` (as opposed to `s.remove(x)`) when `s` is a `set`.
