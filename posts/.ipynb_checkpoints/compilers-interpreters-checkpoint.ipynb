# On Compilers and Interpreters (30 November 2025)

I wrote this quick blog post while reviewing [Python Functions](https://github.com/kedarmhaswade/yakk-2/blob/main/phase-1/day10_pretest.ipynb).

I stumbled upon the following statements as I read the _Functions_ section in _Python In a Nutshell_ by Martelli et.al. (with a few changes for context):

> When the Python **interpreter** (the `python` executable on your computer) encounters a `def` statement in your Python script or REPL, it **compiles** the function body, creating a function object.

This statement subtly describes the dual nature of `python` (the name of a program in an executable form), which is both a compiler and an interpreter. It also prompted me to reexamine _compilation_ and _interpretation_ in the context of programming languages. I had always encountered these terms. I asked myself: **Do I understand these terms and their difference now?** 

This blog post answers that question.

First, let's look up the _contemporary_ dictionary meanings of the two verbs: _compile_ and _interpret_. I have stated the meanings most relevant in a computing context in boldface.

1) **Compile (v.)**: 1) get or gather together (synonym: to accumulate), 2) put together out of existing material (e.g. compile a list), or 3) **use a computer program (a "compiler") to translate source code written in a particular programming language into computer-readable machine code that can be executed**.
2) **Interpret (v.)**: 1) make sense of; assign meaning to (e.g. interpret a message) 2) give rendition of (e.g. a Beethoven sonata) 3) create an image or likeness of (synonym: to represent) 3) **restate words from one language into another language (synonym: to translate)**, or 4) make sense of a language.

Aren't the relevant definitions confusing? Don't they mean the same thing (to translate)? Restating (modified slightly to suit Python, although it doesn't matter) them:

1) **Compile**: Use the `python` executable to translate a complete Python program into the Intel microprocessor's machine code (instructions), which can be later _run_ or _executed_.
2) **Interpret**: Use the `python` executable to translate or restate the words from Python (e.g. an assignment like `x = 20`) into another language like Intel microprocessor's (x86 64) machine code (something [like](http://ref.x86asm.net/coder64.html#xC7) `48 c7 c0 01 00 00 00` encoded in binary) and run it, so that actual physical changes occur at the semiconductor level! This is like how a human interpreter interprets a foreign president's response for the native speakers to _understand_. 

When a program _runs_ on a computer, the CPU simply executes corresponding instructions one by one. A runnable program is a file with 0s and 1s (binary). It is also called an _executable_. The executable then contains a sequence of machine code instructions.

**Thus, the CPU is the ultimate interpreter or executor that runs instructions in its machine code corresponding to any program**. The CPU does _not_ translate the program written in computer's RAM in a 'language' like `48 c7 c0 01 00 00 00` into some other language, but it, like a faithful and unerring servant, "directly makes sense of it" by taking actions in the hardware (thus _interpreting_ it--See meaning #1 of _interpret_ above).

For _our_ program, written in any language other than machine code, to do anything on a physical machine, we need two things:

1) Our program's code somehow **translated** into a binary machine code (instructions for a CPU).
2) A CPU that **interprets** or makes sense of it by taking the instructed actions.

We now have more nuanced definitions:

1) **Compile**: Translate something from one _formal_ language into another. Formal languages, unlike natural languages, have _unambiguous grammars_.
2) **Interpret**: Make sense of (well-defined) instructions in a language by taking actions intended therein.

Until about the 70s (and of course even now), we had a programming language like C in which programmers wrote their programs, a compiler like the `gcc` (which is itself a big program written in C and compiled into a computer's machine code) that _compiled_ those C programs into a computer's machine code, and a machine that _interpreted_ that machine code. A typical workflow of an application programmer was straightforward:

1) *Create* a C program file, `a.c`. This file is not _executable_. It contains the C source code.
2) *Compile* it: For example, `gcc a.c`. This creates an _executable_ file `a.out` which has all the machine instructions. The compiler ensures that `a.c` is _valid_ C code, and `a.out` is valid machine code. Compilers stand on a firm foundation of formal, theoretical, and mathematical principles; it is one of the best-understood parts of Computer Science.
3) *Run* the executable file: `./a.out` which transfers (_loads_) the instructions in computer's memory. We didn't call this step _interpretation_, although the computer makes sense of (interprets) those instructions.

This required that different computers had different compilers, even though the C program file was the same. You could give your little C program, `a.c` (not `a.out`, because it was for _your_ computer), to your friend on a floppy, and s/he would compile it on their computer (which may be different from yours) to, say, `a.exe`, and run it.

What is a compiler like `gcc` then? Clearly, it must be a program, an executable file that has all the instructions for the machine. When run, it takes your C program and produces a corresponding executable. There is, of course, a [bootstrapping problem](https://bootstrappable.org/) here if `gcc` were _not_ available as a sequence of machine instructions. If `gcc` is the name of the executable program that was created by compiling a C (or other language) source program, then we must compile that source program somehow!

For now, let's assume that `gcc` exists.

Later, perhaps in the early 90s, the idea of writing computer programs for a _virtual machine_ came along. We'll not go into the benefits of this idea here.

The advent of virtual machines made the term 'interpreter' more well-known. Now, **a compiler translates programs from a source language into an instruction language, not for any physical machine, but for a virtual machine**. The instructions for the virtual machine were called the _bytecode_. 

A compiler for the virtual machine generates bytecode from a program in some high-level language (such as Python). 

An interpreter interprets (makes sense of) the bytecode (for example, the Python _bytecode_) by transparently translating it into the machine code of a given physical computer. 

Two notable programming languages developed with a virtual machine in mind are Python and Java. Let's consider Java first. An application programmer's workflow now becomes slightly more complicated: 

1) *Create* a Java program file, `A.java`. This file is not _executable_. It contains the Java source code.
2) *Compile* `A.java` using another executable program on your computer, `javac`: This creates another file, called the Java class file, named `A.class`, which has all the bytecode (instructions) **for the Java virtual machine**. `A.class` can not be executed (or run) on any physical computer all by itself.
3) *Run* `A.class` by using yet another executable program on your computer, `java`: This does not create any file. The `java` program simply interprets the bytecode in `A.class` by transparently translating it into the machine code for a given physical computer (CPU), which executes it. The fact remains. **CPU is the ultimate interpreter**. However, there is no separate step, "running the executable file".

In Java, `javac` is the compiler of your Java program, and `java` is the interpreter of the corresponding bytecode. The bytecode is transparently translated into the machine code. The CPU is the ultimate interpreter of the machine code.

For Python, `python` is both the compiler and the interpreter. The two distinct functions are merged into one executable program that compiles and interprets your Python program. The bytecode is transparently translated into the machine code. The CPU is the ultimate interpreter of the machine code.

Consider a simple Python program:

``` python
1 def square(n):
2    return n*n

3 s = square(5)
4 print(s)
```

When the `python` executable encounters a function, the `def` statement (line `1`), it first compiles the function named `square` to the bytecode (remember, `python` is both a compiler and an interpreter) for the Python virtual machine. A subsequent _call_ to that function (via its bytecode corresponding to the line `3`) tells the `python` interpreter to interpret the previously compiled function's bytecode.

The `python` interpreter is a big program written in C, compiled by someone else. You _run_ it on a given computer (just like you ran `a.out` compiled from your `a.c` above). Its source code (at a high level) looks like this:

``` C
for (;;) {
    switch (opcode) {
        /* cases refer to actual bytecodes in a hidden file */
        case LOAD_CONST: ...
        case CALL: ...
        case RETURN_VALUE: ...
        ...
    }
}
```

The situation is subtly different when your code calls a Python built-in function like `print` (line `4`) or `sum`: **Unlike `square`, there is no _bytecode_ available for `print`**. Instead, the interpreter has access to its previously compiled machine code. The `python` interpreter then cedes control to the CPU, which interprets that machine code.

See [this answer on Computer Science StackExchange](https://cs.stackexchange.com/a/148444/10496) for more details.

That's it. 

Advanced operations (such as Ahead-Of-Time_ (AOT), _Just-In-Time_ (JIT) compilation, etc.)  are available to run programs more efficiently. 

Still, the gist is that compilers are fast translators of source code into machine code, *requiring a separate loading/running/execution step on a given machine*. In contrast, interpreters are comparatively slower translators of source code directly into the machine code that *do not require a separate execution step*.
