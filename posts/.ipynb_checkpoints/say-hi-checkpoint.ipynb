This is my first blog post, powered by Quarto and GitHub Workflow. I am not exactly happy because it took half a day to set all this up. But now that it is all set up, I can hopefully produce some readable, version-controlled blog posts. 

## My Workflow
- Keep my phase-wise notebooks (which are expected to be detailed explorations) up-to-date. I decided not to convert them into a website like this one. Instead, they are just _executed notebooks (even with errors to demonstrate something)_ committed to [my GitHub repo](https://github.com/kedarmhaswade/yakk-2).
- Just see the individual notebooks in `phase-1`, `phase-2`, etc. **I link all my phase-star notebooks on my blog using the GitHub source links. Here's a typical link: https://github.com/kedarmhaswade/yakk-2/blob/main/phase-1/day10_pretest.ipynb. It should appear verbatim in a blog post**.
- Only blog posts are rendered on the blog. Executed notebooks are available only on the GitHub repository.
- I have a GitHub workflow that converts and copies notebooks to my blog.
